```python
#Group by
d=pd.read_csv('cars.csv')
```

```python
d.head(4)
```

```python
grouped=d.groupby('fuel')
avg_price=grouped['selling_price'].mean()
```

```python
print(avg_price)
```

```python
#Applying multiple agg function
aggregate=grouped.agg({ 'selling_price':['mean','sum','max'] } )
print(aggregate)
```

```python
#Grouping multiple columns
```

```python
grouped_multi=d.groupby(['brand','fuel'])
```

```python
avg_salary=grouped_multi['selling_price'].mean()
```

```python
print(avg_salary)
```
