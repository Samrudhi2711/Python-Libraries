```python
import pandas as pd
```

```python
df=pd.read_csv("cars.csv")
```

```python
df.head(5)
```

```python
df1=df.groupby('fuel')
```

```python
df1.plot(kind="line",x='fuel',y='selling_price')
```

```python
df1=df.groupby('fuel')['selling_price'].mean()
df1.plot(kind='bar')
```

```python
df1=df.groupby('brand')['km_driven'].mean()
df1.plot(kind='bar')
```

```python
df1=df.groupby('brand')['km_driven'].mean()
df1.plot(kind='barh')
```

```python
df2=df.groupby('brand')['km_driven'].mean()
df2.plot(kind='hist')
```

```python
df1.plot(kind='box')
```

```python
df1.plot(kind='kde')
```
