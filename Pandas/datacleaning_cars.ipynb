```python
import pandas as pd
```

```python
df=pd.read_csv("cars.csv")
df.head(4)
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
#Find the average selling price of cars grouped by brand.
```

```python
brand_group=df.groupby('brand')['selling_price'].mean()
```

```python
brand_group
```

```python
#Find the maximum km_driven for each fuel type.
```

```python
max_km=df.groupby('fuel')['km_driven'].max()
```

```python
max_km
```

```python
#Count how many cars are owned by each owner type.
```

```python
car=df.groupby('owner')['brand'].count()
```

```python
car
```

```python
#Find minimum, maximum, and average selling price for each brand.
```

```python
aggr=df.groupby('brand')['selling_price'].agg(['min','max','mean'])
```

```python
aggr
```

```python
#Show total cars and average km_driven grouped by fuel.
```

```python
totalAvg=df.groupby('fuel').agg(totcar=("brand",'count'),km_avg=("km_driven",'mean') )
```

```python
totalAvg
```

```python
#Find the average price of cars grouped by fuel and owner.
```

```python
avgPrice=df.groupby(['fuel','owner'])['selling_price'].mean()
```

```python
avgPrice
```

```python
#For each brand and fuel type, find the maximum km_driven.
```

```python
max_km=df.groupby(['brand','fuel'])['km_driven'].max()
```
