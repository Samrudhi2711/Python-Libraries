```python
import pandas as pd
```

```python
df=pd.read_csv('sales_data.csv')
```

```python
df.head(6)
```

```python
#Group by brand and fuel
#👉 Example: Find average price for each brand and fuel combination

groupby=df.groupby(['brand','fuel'])['price'].mean()
print(groupby)
```

```python
#1. Average price per brand
avg_price=df.groupby('brand')['price'].mean()
print(avg_price)
```

```python
#Total units sold per fuel type
unit_sold=df.groupby('fuel')['units_sold'].sum()
print(unit_sold)
```

```python
#3. Average price per brand & fuel combination
avg_price_band=df.groupby(['brand','fuel'])['price'].mean()
print(avg_price_band)
```

```python
#Multiple aggregations (price and units_sold together)
agg_data=df.groupby('brand').agg(
    avg_price=('price','mean'),
    max_price=('price','max'),
    avg_unit=('units_sold','mean'),
    max_unit=('units_sold','max')
)

print(agg_data)
```

```python
#Group by year and fuel
agg_data1=df.groupby(['year','fuel']).agg(
    total_unit=('units_sold','sum'),
    avg_price=('price','mean')
)
print(agg_data1)
```

```python
#1. Total units sold by brand
total_unit=df.groupby('brand')['units_sold'].sum()
print(total_unit)
```

```python
#2. Average price by fuel type
a_p=df.groupby('fuel')['price'].mean()
print(a_p)
```

```python
#Brand + Fuel combination (multiple columns groupby)
b_s=df.groupby(['brand','fuel'])['units_sold'].sum()
print(b_s)
```

```python
#Q1. Which brand sold the maximum units each year?
#👉 Find the top-selling brand per year.
max_seal=df.groupby(['brand','year'])['units_sold'].max().reset_index()
max_seal1=max_seal.loc[max_seal.groupby('year')['units_sold'].idxmax()]
print(max_seal1)
```

```python
#Q2. For each fuel type, what is the total revenue (price × units_sold)?
#👉 Compute total sales revenue by fuel.
df['revenue']=df['price']*df['units_sold']
Q2=df.groupby('fuel')['revenue'].sum()
print(Q2)
```

```python
#Q3. What is the average price of cars per brand in each region?
#👉 Compare how brand prices differ across regions.

avg_price=df.groupby(['brand','region'])['price'].mean().unstack()
print(avg_price)
```

```python
#Q5. Which region contributed the highest sales for each fuel type?
#👉 Find dominant region per fuel category.

q5 = df.groupby(['fuel','region'])['units_sold'].sum().reset_index()
q5_top = q5.loc[q5.groupby('fuel')['units_sold'].idxmax()]
print(q5_top)
```
