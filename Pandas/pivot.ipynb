```python
import pandas as pd
```

```python
d=pd.read_csv('deliveries.csv')
```

```python
d.head(3)
```

```python
d.columns
```

```python
mask=d['batsman_runs']==6
six=d[mask]
six.shape
```

```python
pt=six.pivot_table(index='over',columns='batting_team',values='batsman_runs',aggfunc='count')
```

```python
import seaborn as sns
```

```python
sns.heatmap(pt)
```

```python
e=pd.read_csv("Employee.csv")
```

```python
e.head()
```

```python
#Basic pivot table
```

```python
pivot=pd.pivot_table(e,index='Age',values='Salary',aggfunc='mean')
print(pivot)
```

```python
pivot=pd.pivot_table(e,index='Age',values='Salary',aggfunc=['mean','max','sum'])
print(pivot)
```

```python
p_c=pd.pivot_table(e,values='Salary',index='Age',columns='Experience_Years',aggfunc='mean')
```

```python
print(p_c)
sns.heatmap(p_c)
```

```python
pivot_fill=pd.pivot_table(e,values="Salary",index="Age", columns="Experience_Years",fill_value=0)
```

```python
print(pivot_fill)
```
