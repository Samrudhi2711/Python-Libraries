```python
#Encoding TArget Parametr which is species
```

```python
data.species.unique()
```

```python
data.species=data.species.map({'setosa':1, 'versicolor':2, 'virginica':3})
```

```python
sns.heatmap(data.corr(),annot=True)
```

```python
plt.plot(data.petal_width,data.species.map({1:'setosa',2:'versicolor',3:'virginica'}))
plt.plot(data.petal_length,data.species.map({1:'setosa',2:'versicolor',3:'virginica'}))
plt.plot(data.sepal_width,data.species.map({1:'setosa',2:'versicolor',3:'virginica'}))
plt.plot(data.sepal_length,data.species.map({1:'setosa',2:'versicolor',3:'virginica'}))
plt.legend()
plt.xlable("Scale")
plt.ylable("Species")
plt.title("Impact on Species")

plt.show()
```

```python
#Observation
#petal length giving clear indication of species
#range of petal length
#1)Range less than 3cm =Iris setosa
#2)Range 3-5cm=Iris Versicolor
#3)Range more than 5cm =Iris Virginica
```

```python
#Preparing input and target data
```

```python
input_parameter = data.drop("species",axis=1)
target_parameter = data.species
```

```python
input_parameter.shape
```

```python
target_parameter.shape
```

```python
#standard Scaling data
```

```python
sc=StandardScaler()
input_parameter =sc.fit_transform(input_parameter)
```

```python
#Spliying data to train test
```

```python
input_train, input_test, target_train, target_test = train_test_split(input_parameter, target_parameter, test_size=0.2, random_state=42)
```

```python
#algoritham Decision Tree
```

```python
dt=DecisionTreeClassifier()
dt.fit(input_train,target_train)
dtp=dt.predict(input_test)
dta=accuracy_score(dtp,target_test)
print("percent acciracy =",round(dta*100,2),"%")
```

```python
#Algoritham- KNeighbours
```

```python
kn=KNeighborsClassifier(n_neighbors=9)
kn.fit(input_train,target_train)
knp=kn.predict(input_test)
kna=accuracy_score(knp,target_test)
print("Precent Accuracy =",round(kna*100,2),"%")
```

```python
#Algoritham-Random Forest
```

```python
rf=RandomForestClassifier(n_estimators=500)
rf.fit(input_train,target_train)
rfp=rf.predict(input_test)
rfp=accuracy_score(rfp,target_test)
print("percent Accuracy =",round(rfp*100,2),"%")
```
