```python
import requests
url = "https://en.wikipedia.org/wiki/List_of_countries_by_population_(United_Nations)"
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 "
                  "(KHTML, like Gecko) Chrome/116.0.0.0 Safari/537.36"
}
response = requests.get(url, headers=headers)

tables = pd.read_html(response.text)
print(len(tables))        # number of tables found on the page
print(tables[0].head())  
```

```python
#pd.DataFrame(dict) is one of the easiest ways to make a DataFrame.
#Dictionary keys → columns, values → data.
#You can customize row index and column order.
```

```python
data={
   "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 28],
    "City": ["Pune", "Mumbai", "Delhi"]
}
df=pd.DataFrame(data)
df
```

```python
#Custom row index
```

```python
df=pd.DataFrame(data,index=['a','b','c'])
df
```

```python
#Selecting column order
```

```python
df=pd.DataFrame(data,columns=['City','Name','Age'])
df
```

```python
#df.to_csv()
```

```python
df.to_csv('people.csv',index=False)
df=pd.read_csv("people.csv")
df
```

```python
df.to_excel("people.xlsx")
df
```

```python
df.to_json("people.json")
df
```
